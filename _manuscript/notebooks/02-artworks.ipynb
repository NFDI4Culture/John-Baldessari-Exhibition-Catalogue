A Juypter Notebook for retreiving artwork information from Wikidata item [https://www.wikidata.org/wiki/Q141747732](https://www.wikidata.org/wiki/Q141747732) | Exhibition title: while something is happening here, something else is happening there : works 1988 - 1999 (Exhibition)

```python
from urllib.parse import urlencode
from urllib.request import Request, urlopen
import json

exhibition_id = "Q141747732"
query = f"""
SELECT ?item ?itemLabel WHERE {{
  wd:{exhibition_id} wdt:P527 ?item .
  SERVICE wikibase:label {{ bd:serviceParam wikibase:language "[AUTO_LANGUAGE],en". }}
}}
ORDER BY ?itemLabel
LIMIT 100
"""

endpoint = "https://query.wikidata.org/sparql"
params = urlencode({"query": query, "format": "json"})
url = f"{endpoint}?{params}"

request = Request(
    url,
    headers={
        "Accept": "application/sparql-results+json",
        "User-Agent": "loe-posit-notebook/1.0 (contact: simon.worthington@tib.eu)",
    },
)

with urlopen(request, timeout=60) as response:
    data = json.loads(response.read().decode("utf-8"))

rows = []
for b in data["results"]["bindings"]:
    item_id = b["item"]["value"].rsplit("/", 1)[-1]
    label = b.get("itemLabel", {}).get("value", item_id)
    rows.append({"item_id": item_id, "item": label})

rows.sort(key=lambda r: r["item"].lower())

print(f"Items listed as parts of exhibition {exhibition_id}")
if rows:
    for r in rows:
        print(f"{r['item_id']} | {r['item']}")
else:
    print("No items were found as part of this exhibition.")

print(f"\nTotal items: {len(rows)}")
```

```python
from urllib.parse import urlencode
from urllib.request import Request, urlopen
import json

exhibition_id = "Q141747732"
query = f"""
SELECT ?artwork ?artworkLabel WHERE {{
  ?artwork wdt:P31 wd:Q3305213 ;
           wdt:P608 wd:{exhibition_id} .
  SERVICE wikibase:label {{ bd:serviceParam wikibase:language "[AUTO_LANGUAGE],en". }}
}}
ORDER BY ?artworkLabel
LIMIT 100
"""

endpoint = "https://query.wikidata.org/sparql"
params = urlencode({"query": query, "format": "json"})
url = f"{endpoint}?{params}"

request = Request(
    url,
    headers={
        "Accept": "application/sparql-results+json",
        "User-Agent": "loe-posit-notebook/1.0 (contact: simon.worthington@tib.eu)",
    },
)

with urlopen(request, timeout=60) as response:
    data = json.loads(response.read().decode("utf-8"))

rows = []
for b in data["results"]["bindings"]:
    artwork_id = b["artwork"]["value"].rsplit("/", 1)[-1]
    label = b.get("artworkLabel", {}).get("value", artwork_id)
    rows.append({"artwork_id": artwork_id, "artwork": label})

rows.sort(key=lambda r: r["artwork"].lower())

print(f"Artworks associated with exhibition {exhibition_id}")
if rows:
    for r in rows:
        print(f"{r['artwork_id']} | {r['artwork']}")
else:
    print("No artworks were found for this exhibition.")

print(f"\nTotal artworks: {len(rows)}")
```
