# Curators’ Guide

Step-by-step instructions for creating public records for art works in an art exhibitions and producing a catalogue publication

## Purpose

The purpose of this guide is to provide curators with a practical approach to making public records of art works in an exhibition.

This can be done from the following sources:

- Retrospective - from exhibition book catalogues, data records, and spreadsheets.
- Current - from ongoing exhibition records, museum databases, and digital catalogues.

The process is aligned with relevant cultural heritage data standards - primarily [CIDOC CRM](https://cidoc-crm.org/), its simplified derivative [Linked Art](https://linked.art/), [LIDO](https://www.lido-schema.org/), [Art & Architecture Thesaurus® Online - AAT](https://www.getty.edu/research/tools/vocabularies/aat/), [InConClass](https://www.inconclass.org/), and the many community-built [Wikidata schemas](https://www.wikidata.org/wiki/Wikidata:WikiProject_Arts) and related property guidance for exhibition data.

- WikiProject Visual Arts: <https://en.wikipedia.org/wiki/Wikipedia:WikiProject_Visual_arts>
- Wikiproject Exhibitions: <https://www.wikidata.org/wiki/Wikidata:WikiProject_Exhibitions>

The results of the process are:

- Public records of artworks and exhibitions in a structured, machine-readable format, and compliance with cultural heritage data standards
- A foundation for creating exhibition catalogues from public data

### Learning point

- How to make Linked Open Data entries for artworks and exhibitions
- Understand data standards for art and exhibitions
- Creating an exhibition book catalogue from public data

The guide has four parts:

1.  Making Linked Open Data entries in Wikidata
2.  Using Jupyter Notebooks
3.  Data standards for art exhibitions
4.  Publishing an exhibition catalogue

## 1. Making Linked Open Data entries in Wikidata

### Wikidata

**Exercise #1:** Make an artwork entry in Wikidata.

Select an artwork from the list. The full artwork list from the exhibition catalogue OCR scan can be found in [catalogue-baldessari.qmd](01-baldessari-ocr-data.qmd). The ones with Q numbers at start have already been added.

Here is an example artwork in Wikidata <https://www.wikidata.org/wiki/Q141749370> (John Baldessari, “Duck Pond Bar, National City, California”, 1967)

### How to add an item to Wikidata

Login or create an account at <https://www.wikidata.org/> (Wikidata main page)

**IMPORTANT! Use the Artwork ER model data field list below!**

1.  Sign in to Wikidata and choose the “Create a new item” option.
2.  Give the artwork a clear label `'Artwork Title'` and a short description.
3.  Add the key statements for the work: title, type, date, materials, dimensions, and exhibition history.
4.  Use the relevant Wikidata properties, such as P1476 for title, P31 for instance of, and P571 for inception date.
5.  Save the item and check that the new record is consistent with the existing data model.

## Artwork ER model

Add data for these fields.

**IMPORTANT! Use *work of art* Q838948 for the instance of field “Instance of” P31, and use the title: *while something is happening here, something else is happening there : works 1988 - 1999* Q141747732 for the “Exhibition history” P608**

| No. | Field | Data type | Wikidata | CIDOC CRM |
|---------|------------|------------------|-----------------|------------------|
| 1 | **Instance of** | Item | P31 | E22_Man-Made_Object |
| 2 | **Title** | Monolingual text | P1476 | P102_has_title (E35_Title) |
| 3 | **Creator** | Item | P170 | P14_carried_out_by (E39_Actor) |
| 4 | **Inception** | Point in time | P571 | P108_has_produced (E65_Creation) |
| 5 | **Materials** | Item | P186 | P45_consists_of (E57_Material) |
| 6 | **Width** | Quantity | P2049 | P43_has_dimension (E54_Dimension) |
| 7 | **Height** | Quantity | P2048 | P43_has_dimension (E54_Dimension) |
| 8 | **Exhibition history** | Item | P608 | E7_Activity (exhibition event) |
| 9 | **Commons category** | String | P373 | \- |

#### Notes when entering statements

- 1.  **P31 (instance of)**: usually a single main value is enough; add only the most specific artwork type and avoid conflicting classes, e.g., painting, photograph, etc. [Q838948](https://www.wikidata.org/wiki/Q838948) (work of art)
- 1.  **P1476 (title)**: use the preferred title in its original language, add language qualifiers (en) or a translated title as a second statement. The field for setting language appears ontop of the title input box.
- 1.  **P170 (creator)**: if there are multiple contributors, add separate creator statements and consider a role qualifier such as P3831 to say whether the person was artist, photographer, or designer. Add: John Baldessari (this will link to artist Wikidata entry)
- 1.  **P571 (inception)**: record the date as precisely as possible; if only a period is known, use a time range or a general date instead of a guessed exact day.
- 1.  **P186 (made from materials)**: each material is usually added as its own statement and should point to a Wikidata item such as oil, canvas, bronze, or marble.
- 1.  / 7. **P2048/P2049 (height/width)**: these are quantity statements and must include a unit (for example cm or m) and a value; the unit should be consistent across related works.
- 1.  **P608 (exhibition history)**: this is an exhibition-item relationship, so the statement value should be an exhibition item; if dates are known, add them as qualifiers or separate event context. Use: [Q141747732](https://www.wikidata.org/wiki/Q141747732)
- 1.  **P373 (Commons category)**: enter only the category name without the `Category:` prefix. Type it in with spaces (John Baldessari). See: [Category:John_Baldessari](https://commons.wikimedia.org/wiki/Category:John_Baldessari)
- General rule: Review all statements for accuracy and completeness before finalizing the entry.

## 3. Entity-relationship models for art exhibitions

**Exercise 3#:** Discuss data modelling and building Entity-Relationship models. Using LLM chat to simplify use of Linked Open Data and the world of strcutured data, e.g., <https://gemini.google.com/>. Recommendation is ‘Keep it Simple’, first go bottom up - look at what the thing is, then top down - see how the bottom up process maps to the formal world.

Schemas and key concepts: <https://en.wikiversity.org/wiki/BIM-126-02-Data-Science-Linked-Open-Exhibition#Schemas_and_key_concepts>

The project currently includes the following entity-relationship models:

- [artist-datamodel.qmd](artist-datamodel.qmd): describes the artist as an agent with identity, role, and provenance information.
- [item-in-exhibition-datamodel.qmd](item-in-exhibition-datamodel.qmd): models the relationship between an artwork and the exhibition event in which it appears.
- [exhibition-datamodel.qmd](exhibition-datamodel.qmd): defines the exhibition as an event with title, dates, venue, and institutional context.
- [dnb-catalogue-datamodel.qmd](dnb-catalogue-datamodel.qmd): represents the catalogue record and its bibliographic metadata as a source for exhibition information.
