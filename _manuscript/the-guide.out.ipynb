# Curators’ Guide

Step-by-step instructions for using the Linked Open Exhibition model to create open access catalogues records for art exhibitions

All materials here are open-licenced unless otherwise stated and follow FAIR Data principles and follow UNESCO Open Science recommendations.

## Purpose

The purpose of this guide is to provide curators with a structured approach to creating open access catalogue records for art exhibitions using the Linked Open Exhibition model. It covers the process from entering data into Wikidata and Wikibase, to using computational tools for managing and publishing exhibition information.

The cataloguing process is aligned to CIDOC CRM and other relevant cultural heritage data standards.

### Learning point

- How to make Linked Open Data entries for artworks and exhibitions in Wikidata and Wikibase
- Use computational publishing tools to manage and publish exhibition information, e.g., Jupyter Notebooks, Quarto, and open-weight/source LLMs
- Understand the entity-relationship models for art exhibitions: e.g., CIDOC CRM and Linked Art

The guide has four parts:

1.  Making Linked Open Data entries in Wikidata and Wikibase
2.  Using Jupyter Notebooks and MyBinder
3.  Entity-relationship models for art exhibitions
4.  Quarto publishing for exhibition catalogues

## 1. Making Linked Open Data entries in Wikidata and Wikibase

### Wikidata

The exercise is to make an artwork entry in Wikidata from the list generated by the OCR scan. The full artwork list from the exhibition catalogue OCR scan can be found in [catalogue-baldessari.qmd](catalogue-baldessari.qmd).

Use the entity-relationship structure defined in [item-in-exhibition-datamodel.qmd](item-in-exhibition-datamodel.qmd) as the reference model for describing artworks and their exhibition context in Wikidata and Wikibase.

Here is an example artwork https://www.wikidata.org/wiki/Q141749370 (John Baldessari, “Duck Pond Bar, National City, California”, 1967)

### How to add an item to Wikidata

https://www.wikidata.org/ (Wikidata main page)

1.  Sign in to Wikidata and choose the “Create a new item” option.
2.  Give the artwork a clear label and, if needed, a short description.
3.  Add the key statements for the work: title, type, date, materials, dimensions, and exhibition context.
4.  Use the relevant Wikidata properties, such as P1476 for title, P31 for instance of, and P571 for inception date.
5.  Save the item and check that the new record is consistent with the existing data model.

There are the minimum Statements to add about an item:

NB Exhibition history should be added as an item linking to the specific exhibition event, rather than just a text description. ‘while something is happening here, something else is happening there : works 1988 - 1999’ https://www.wikidata.org/wiki/Q141747732

#### Notes when entering statements

- 1.  **P31 (instance of)**: usually a single main value is enough; add only the most specific artwork type and avoid conflicting classes.
- 1.  **P1476 (title)**: use the preferred title in its original language, and if needed add language qualifiers or a translated title as a second statement.
- 1.  **P170 (creator)**: if there are multiple contributors, add separate creator statements and consider a role qualifier such as P3831 to say whether the person was artist, photographer, or designer.
- 1.  **P571 (inception)**: record the date as precisely as possible; if only a period is known, use a time range or a general date instead of a guessed exact day.
- 1.  **P186 (materials)**: each material is usually added as its own statement and should point to a Wikidata item such as oil, canvas, bronze, or marble.
- 6-7. **P2048/P2049 (height/width)**: these are quantity statements and must include a unit (for example cm or m) and a value; the unit should be consistent across related works.
- 1.  **P608 (exhibition history)**: this is an exhibition-item relationship, so the statement value should be an exhibition item; if dates are known, add them as qualifiers or separate event context.
- 1.  **P373 (Commons category)**: enter only the category name without the `Category:` prefix.
- General rule: for all 1-9, add a reference for the source of the statement, such as the catalogue entry, the OCR transcription, or a museum record, since these values are derived from catalogue evidence.

## 2. Using Jupyter Notebooks and MyBinder

A Jupyter Notebook is an interactive document that combines narrative text, executable code, and output in a single file. It is especially useful for cleaning and transforming catalogue data, checking OCR results, and making the workflow transparent and reproducible. In this project, the artworks example is [notebooks/02-artworks.ipynb](https://github.com/NFDI4Culture/John-Baldessari-Exhibition-Catalogue/blob/main/notebooks/02-artworks.ipynb), which shows artworks related to an exhibition as entered on Wikidata.

## 3. Entity-relationship models for art exhibitions

The project currently includes the following entity-relationship models:

- [artist-datamodel.qmd](artist-datamodel.qmd): describes the artist as an agent with identity, role, and provenance information.
- [item-in-exhibition-datamodel.qmd](item-in-exhibition-datamodel.qmd): models the relationship between an artwork and the exhibition event in which it appears.
- [exhibition-datamodel.qmd](exhibition-datamodel.qmd): defines the exhibition as an event with title, dates, venue, and institutional context.
- [dnb-catalogue-datamodel.qmd](dnb-catalogue-datamodel.qmd): represents the catalogue record and its bibliographic metadata as a source for exhibition information.
